```python
# Q2
A_i = int(input())
A_j = int(input())

B_i = int(input())
B_j = int(input())

A = [[int(input()) for j in range(A_j)] for i in range(A_i)]
B = [[int(input()) for j in range(B_j)] for i in range(B_i)]

if A_j == B_i:
    for i in range(A_i):
        for j in range(B_j):
            s = 0
            for k in range(A_j):
                s += A[i][k] * B[k][j]
            print(s, end=" ")
        print()
else:
    print("NOT MULTIPLICABLE")

```

```python
# Alternate solution for Q2

A = [list(map(int, row.strip().split())) for row in input().split(";")] # 1 2 3; 4 5 6; 7 8 9
B = [list(map(int, row.strip().split())) for row in input().split(";")] # 9 8 7; 6 5 4; 3 2 1

A_i = len(A)
A_j = len(A[0])

B_i = len(B)
B_j = len(B[0])

if A_j == B_i:
    for i in range(A_i):
        for j in range(B_j):
            s = 0
            for k in range(A_j):
                s += A[i][k] * B[k][j]
            print(s, end=" ")
        print()
else:
    print("NOT MULTIPLICABLE")

```

```python
# Q4

A = [list(map(int, row.strip().split())) for row in input().split(";")] # 1 2 3; 4 5 6; 7 8 9
A_T = [[A[j][i] for j in range(len(A))] for i in range(len(A[0]))] # Transpose
print(A_T)
```
